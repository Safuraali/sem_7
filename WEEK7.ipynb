```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
# Create dataset
data = {'Size_cm': [7, 7.5, 6.8, 18, 19, 17.5, 8.5, 9, 8.2, 7.2],
    'Color_Intensity': [0.8, 0.75, 0.82, 0.6, 0.58, 0.62, 0.9, 0.88, 0.92, 0.78],
    'FruitType': ['Apple', 'Apple', 'Apple', 'Banana', 'Banana',
                  'Banana', 'Orange', 'Orange', 'Orange', 'Apple']}
df = pd.DataFrame(data)
# Separate features and target
X = df[['Size_cm', 'Color_Intensity']]
y = df['FruitType']
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
# Train KNN model
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)
# Make predictions
y_pred = model.predict(X_test)
# Calculate accuracy
print("Predictions:", y_pred)
print("Actual Values:", list(y_test))
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
# Create dataset
temp = np.array([10, 15, 20, 25, 30, 12, 18, 23, 28, 32]).reshape(-1, 1)
kwh = np.array([50, 60, 75, 90, 105, 55, 70, 85, 95, 110])
# Split data
X_train, X_test, y_train, y_test = train_test_split(temp, kwh, test_size=0.3, random_state=42)
# Train Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)
# Make predictions
y_pred = model.predict(X_test)
# Calculate MSE
print("Predictions:", y_pred)
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
# Visualize data and regression line
plt.scatter(X_train, y_train, color='blue', label='Actual Data')
plt.plot(X_train, model.predict(X_train), color='red', label='Regression Line')
plt.xlabel("Temperature (°C)")
plt.ylabel("Electricity Consumption (kWh)")
plt.title("Temperature vs Electricity Consumption")
plt.legend()
plt.show()
```
