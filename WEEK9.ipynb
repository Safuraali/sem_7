```python
import numpy as np
# AND gate inputs and outputs
X_and = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_and = np.array([0, 0, 0, 1])
# Initialize weights and bias
np.random.seed(42)
weights = np.random.rand(2) * 0.1
bias = np.random.rand() * 0.1
learning_rate = 0.1
epochs = 20
# Step activation function
def step_function(z):
    return 1 if z >= 0 else 0
# Training
for epoch in range(epochs):
    for x, y_true in zip(X_and, y_and):
        z = np.dot(x, weights) + bias
        prediction = step_function(z)
        error = y_true - prediction
        weights = weights + learning_rate * error * x
        bias = bias + learning_rate * error
# Testing
print("AND Gate Results:")
for x in X_and:
    z = np.dot(x, weights) + bias
    print(x, "=", step_function(z))
```

```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
# Shape dataset
X_shapes = np.array([
    [4, 1.0], [4, 1.05], [4, 0.95],
    [0, 1.0], [0, 1.1], [0, 0.9],
    [4, 0.5], [0, 2.0]])
# 0 = Square, 1 = Circle
y_shapes = np.array([0, 0, 0, 1, 1, 1, 0, 1])
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(X_shapes, y_shapes, test_size=0.3, random_state=42)
# Create MLP model
model = MLPClassifier(hidden_layer_sizes=(5,),max_iter=500,random_state=42,solver='adam')
# Train model
model.fit(X_train, y_train)
# Prediction
y_pred = model.predict(X_test)
print("Predictions:", y_pred)
print("Actual Values:", y_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
# Load MNIST dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
# Normalize images
X_train = X_train / 255.0
X_test = X_test / 255.0
# Create neural network
model = Sequential([Flatten(input_shape=(28, 28)),Dense(128, activation='relu'), Dense(10, activation='softmax')])
# Compile model
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
# Train model
model.fit(X_train, y_train, epochs=5)
# Test model
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Accuracy:", accuracy)
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-10, 10, 100)
# Activation functions
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)
# Plot graphs
plt.plot(x, sigmoid, label='Sigmoid')
plt.plot(x, tanh, label='Tanh')
plt.plot(x, relu, label='ReLU')
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Activation Functions Comparison")
plt.legend()
plt.grid()
plt.show()
```
